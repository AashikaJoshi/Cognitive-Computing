```python
a=1024170209
l=[]
while(a>0):
    l.append((a%10)*10)
    a//=10
l.reverse()
l
```

```python
scores = tuple(l[:8])
scores
```

```python
h = max(scores)
h_index = scores.index(h)

l = min(scores)
l_count = scores.count(l)

print("Highest score:", h)
print("Index of highest score:", h_index)
print("Lowest score:", l)
print("Number of times lowest appears:", l_count)
```

Q3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating
your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed
produces a different list, so this cannot be copied from a classmate).
i. Generate a list of 100 random numbers between 100 and 900 (inclusive).

```python
import random

random.seed(1024170209)
numbers = [random.randint(100, 900) for i in range(100)]
print(numbers)
```

ii. Count and print all odd numbers in the list.

```python
count=0
for i in numbers:
    if i%2!=0:
        count+=1
        print(i)
print("Count of odd numbers:",count)
```

iii. Count and print all even numbers in the list.

```python
count=0
for i in numbers:
    if i%2==0:
        count+=1
        print(i)
print("Count of even numbers:",count)
```

iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers
found using list comprehension.

```python
def isPrime(n):
    if n < 2:
        return False

    for i in range(2, int(n/2) + 1):
        if n % i == 0:
            return False

    return True

prime_numbers = []
for x in numbers:
    if isPrime(x):
        prime_numbers.append(x)

print(prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
```

v. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
m = max(numbers, key=(numbers.count))
m,numbers.count(m)
```

Q4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for
each digit} and set B = {digit × 9 for each digit}. Print both sets.
vi. Find the union of A and B — all unique values across both sets.

```python
digits = [1, 0, 2, 4, 1, 7, 0, 2]
A = {d * 7 for d in digits}
B = {d * 9 for d in digits}

print("A =", A)
print("B =", B)

print("Union =", A.union(B))
```

vii. Find the intersection of A and B — values common to both.

```python
print("Intersection =", A.intersection(B))
```

viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this
differs from symmetric_difference().

```python
print("A - B =", A.difference(B))
print("B - A =", B.difference(A))
# difference() gives elements present in one set but not the other.
# symmetric_difference() gives elements unique to both sets together.
```

ix. Find the symmetric difference of A and B.

```python
print("Symmetric Difference =", A.symmetric_difference(B))
```
