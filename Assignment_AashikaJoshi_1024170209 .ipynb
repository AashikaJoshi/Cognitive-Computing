```python
def odd_sum(n):
    return sum(range(1, n + 1, 2))
print("Odd Sum till 87 =", odd_sum(87))
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
def prime_sum(n):
    return sum(i for i in range(2, n + 1) if is_prime(i))

print("Prime Sum till 101 =", prime_sum(101))
```
