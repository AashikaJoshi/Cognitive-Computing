b) Find the most frequent value and their indice(s) in the following arrays:
i. x = np.array([1,2,3,4,5,1,2,1,1,1])

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly
spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array,
shape, total elements, the data type of each element and total number of bytes consumed
by the array. Find the transpose of this array using reshape() attribute. Can we do the same
with T attribute?

```python
Aashika = np.linspace(10, 100, 25)

print("Array:", Aashika)
print("Dimensions:", Aashika.ndim)
print("Shape:", Aashika.shape)
print("Total elements:", Aashika.size)
print("Data type:", Aashika.dtype)
print("Total bytes:", Aashika.nbytes)

transpose_reshape = Aashika.reshape(1, 25)
print("Transpose using reshape():")
print(transpose_reshape)

transpose_T = Aashika.T
print("Using T attribute:")
print(transpose_T)

print("For a 1-D array, T does not change its shape.")
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>>
with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean,
median, max, min, unique elements. Reshape the array to four rows and three columns and
name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three
columns and name it as resized_ ucs420_<your_name>>.

```python
ucs420_Aashika = np.array([[10, 20, 30, 40],
                             [50, 60, 70, 80],
                             [90, 15, 20, 35]])

print("Mean:", np.mean(ucs420_Aashika))
print("Median:", np.median(ucs420_Aashika))
print("Maximum:", np.max(ucs420_Aashika))
print("Minimum:", np.min(ucs420_Aashika))
print("Unique elements:", np.unique(ucs420_Aashika))

reshaped_ucs420_Aashika = ucs420_Aashika.reshape(4, 3)
print("Reshaped array:")
print(reshaped_ucs420_Aashika)

resized_ucs420_Aashika = np.resize(ucs420_Aashika, (2, 3))
print("Resized array:")
print(resized_ucs420_Aashika)
```
